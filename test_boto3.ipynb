```python
import boto3

# Se connecter avec le profil 's3-user'
session = boto3.Session(profile_name="s3-user")

# Initialiser le client S3 avec la session
s3_client = session.client("s3")

# Bucket et chemin du fichier
S3_BUCKET_NAME = "dsba-mlops-project-bucket"
FILE_NAME = "car_price_dataset.csv"
S3_PATH = "car_price_dataset/car_price_dataset.csv"

# Upload du fichier
s3_client.upload_file(FILE_NAME, S3_BUCKET_NAME, S3_PATH)

print(f"✅ Upload réussi : s3://{S3_BUCKET_NAME}/{S3_PATH}")
```
